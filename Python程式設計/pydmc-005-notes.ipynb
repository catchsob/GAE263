````python
from random import choice

allowed = ['剪刀', '石頭', '布']

user_win = computer_win = 0

while user_win < 2 and computer_win < 2:
    computer = choice(allowed)
    print(f'computer: {computer}')
    
    user = input(f'請輸入{'、'.join(allowed)}')
    if user in allowed:
        print(f'user: {user}')
        
        if ((user == '布' and computer == '石頭') or
            (user == '石頭' and computer == '剪刀') or
            (user == '剪刀' and computer == '布')):   # user win
            print('user win')
            user_win += 1
        elif user == computer:  # tie
            print('tie')
        else:  # user lose
            print('user lose')
            computer_win += 1
    else:
        print(f'bad input')

print('恭喜 user' if user_win == 2 else '恭喜電腦')
````

````python
import random

NUMBERS = ['2', 'A', 'K', 'Q', 'J', '10', '9', '8', '7', '6', '5', '4', '3']
SUITS = ['SPADE', 'HEART', 'DIAMOND', 'CLUB']

#prepare 52 cards
cards = [(s, n) for s in SUITS for n in NUMBERS]

# shuffle
random.shuffle(cards)

def deal(n):
    global cards
    _cards = cards[:n]  # keep first n-cards
    cards = cards[n:]  # refresh cards 
    return _cards
````

````python
nums = [5, 2, 7, 1, 3]

n = len(nums)

for j in range(n-1):
    print(j, 'before', nums)
    for i in range(n-1-j):   # n-1-0, n-1-1, n-1-2, n-1-3 
        if nums[i] > nums[i+1]:
            nums[i], nums[i+1] = nums[i+1], nums[i]
            print(j, 'swap  ', nums)
    print(j, 'after ', nums)
````

````python
def bubble(nums):
    n = len(nums)
    for j in range(n-1):
        for i in range(n-1-j):
            if nums[i] > nums[i+1]:
                nums[i], nums[i+1] = nums[i+1], nums[i]
````
